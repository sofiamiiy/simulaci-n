**# Método de la Transformada Inversa**

Sea

**\(U \sim \mathcal{U}(0,1)\).**

El método de la transformada inversa consiste en generar una variable aleatoria **\(X\)** mediante:

**
\[
X = F^{-1}(U)
\]**

donde

**\(F(x)\) **

es la función de distribución acumulada de la variable aleatoria deseada


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import weibull_min, gumbel_r, cauchy, logistic, pareto

np.random.seed(123)

n = 10000
```

** (a) Distribución Weibull \(Wei(\alpha,1)\)**

La función de distribución acumulada es:

\[
F(x)=1-e^{-x^\alpha}, \qquad x>0
\]

Aplicando el método de la transformada inversa:

** \[
U = 1-e^{-x^\alpha}
\]

Despejando \(x\):

\[
e^{-x^\alpha}=1-U
\]

\[
-x^\alpha = \ln(1-U)
\]

\[
x = \left[-\ln(1-U)\right]^{1/\alpha}
\]

Como \(1-U \sim U\), se usa:

\[
X = (-\ln U)^{1/\alpha}
\]

```python
#weibull

alpha = 2

U = np.random.uniform(0,1,n)

X_weibull = (-np.log(U))**(1/alpha)

print("MEDIA:", np.mean(X_weibull))
print("VARIANZA:", np.var(X_weibull))
print("DESV. EST.:", np.std(X_weibull))

#histograma
x = np.linspace(0,3,500)

plt.figure(figsize=(8,5))
plt.hist(X_weibull, bins=50, density=True, alpha=0.6)

plt.plot(
    x,
    weibull_min.pdf(x, c=alpha),
    linewidth=2
)

plt.title("Histograma Weibull")
plt.xlabel("x")
plt.ylabel("densidad")
plt.grid()
plt.show()

#distribucion acumulada
X_sorted = np.sort(X_weibull)
F_emp = np.arange(1,n+1)/n

plt.figure(figsize=(8,5))
plt.plot(X_sorted, F_emp, label="Empírica")

plt.plot(
    x,
    weibull_min.cdf(x, c=alpha),
    linewidth=2,
    label="Teórica"
)

plt.title("FDA Weibull")
plt.xlabel("x")
plt.ylabel("F(x)")
plt.legend()
plt.grid()
plt.show()
```

# (b) Distribución Gumbel \(Gu(0,1)\)

La FDA es:

\[
F(x)=e^{-e^{-x}}
\]

Aplicando transformada inversa:

\[
U=e^{-e^{-x}}
\]

Tomando logaritmos:

\[
\ln(U)=-e^{-x}
\]

\[
-\ln(U)=e^{-x}
\]

Tomando nuevamente logaritmos:

\[
x=-\ln(-\ln(U))
\]

Por lo tanto:

\[
X=-\ln(-\ln(U))
\]

```python
#gumbel
U = np.random.uniform(0,1,n)

X_gumbel = -np.log(-np.log(U))

print("MEDIA:", np.mean(X_gumbel))
print("VARIANZA:", np.var(X_gumbel))
print("DESV. EST.:", np.std(X_gumbel))

#histograma
x = np.linspace(-4,8,500)

plt.figure(figsize=(8,5))
plt.hist(X_gumbel, bins=50, density=True, alpha=0.6)

plt.plot(
    x,
    gumbel_r.pdf(x),
    linewidth=2
)

plt.title("Histograma Gumbel")
plt.xlabel("x")
plt.ylabel("densidad")
plt.grid()
plt.show()

#distribución acumulada
X_sorted = np.sort(X_gumbel)
F_emp = np.arange(1,n+1)/n

plt.figure(figsize=(8,5))
plt.plot(X_sorted, F_emp, label="Empírica")

plt.plot(
    x,
    gumbel_r.cdf(x),
    linewidth=2,
    label="Teórica"
)

plt.title("FDA Gumbel")
plt.xlabel("x")
plt.ylabel("F(x)")
plt.legend()
plt.grid()
plt.show()
```

# (c) Distribución Cauchy \(C(0,1)\)

La FDA es:

\[
F(x)=\frac{1}{\pi}\arctan(x)+\frac{1}{2}
\]

Aplicando transformada inversa:

\[
U=\frac{1}{\pi}\arctan(x)+\frac{1}{2}
\]

\[
\arctan(x)=\pi\left(U-\frac12\right)
\]

\[
x=\tan\left[\pi\left(U-\frac12\right)\right]
\]

Entonces:

\[
X=\tan\left[\pi\left(U-\frac12\right)\right]
\]

```python
#cauchy
U = np.random.uniform(0,1,n)

X_cauchy = np.tan(np.pi*(U-0.5))

print("MEDIA:", np.mean(X_cauchy))
print("VARIANZA:", np.var(X_cauchy))
print("DESV. EST.:", np.std(X_cauchy))

#histograma
x = np.linspace(-10,10,500)

plt.figure(figsize=(8,5))
plt.hist(X_cauchy, bins=100, density=True, alpha=0.6)

plt.plot(
    x,
    cauchy.pdf(x),
    linewidth=2
)

plt.title("Histograma Cauchy")
plt.xlabel("x")
plt.ylabel("densidad")
plt.grid()
plt.show()

#distribución acumulada
X_sorted = np.sort(X_cauchy)
F_emp = np.arange(1,n+1)/n

plt.figure(figsize=(8,5))
plt.plot(X_sorted, F_emp, label="Empírica")

plt.plot(
    x,
    cauchy.cdf(x),
    linewidth=2,
    label="Teórica"
)

plt.title("FDA Cauchy")
plt.xlabel("x")
plt.ylabel("F(x)")
plt.legend()
plt.grid()
plt.show()
```

# (d) Distribución Logística \(L(0,1)\)

La FDA es:

\[
F(x)=\frac{1}{1+e^{-x}}
\]

Aplicando transformada inversa:

\[
U=\frac{1}{1+e^{-x}}
\]

Despejando:

\[
\frac{1}{U}=1+e^{-x}
\]

\[
e^{-x}=\frac{1-U}{U}
\]

\[
x=-\ln\left(\frac{1-U}{U}\right)
\]

Equivalentemente:

\[
X=\ln\left(\frac{U}{1-U}\right)
\]


```python
#logística
U = np.random.uniform(0,1,n)

X_logistic = np.log(U/(1-U))

print("MEDIA:", np.mean(X_logistic))
print("VARIANZA:", np.var(X_logistic))
print("DESV. EST.:", np.std(X_logistic))

#histograma
x = np.linspace(-8,8,500)

plt.figure(figsize=(8,5))
plt.hist(X_logistic, bins=50, density=True, alpha=0.6)

plt.plot(
    x,
    logistic.pdf(x),
    linewidth=2
)

plt.title("Histograma Logística")
plt.xlabel("x")
plt.ylabel("densidad")
plt.grid()
plt.show()

#distribución acumulada
X_sorted = np.sort(X_logistic)
F_emp = np.arange(1,n+1)/n

plt.figure(figsize=(8,5))
plt.plot(X_sorted, F_emp, label="Empírica")

plt.plot(
    x,
    logistic.cdf(x),
    linewidth=2,
    label="Teórica"
)

plt.title("FDA Logística")
plt.xlabel("x")
plt.ylabel("F(x)")
plt.legend()
plt.grid()
plt.show()

```

# (e) Distribución Pareto \(Par(\alpha)\)

La FDA es:

\[
F(x)=1-x^{-\alpha}, \qquad x\ge 1
\]

Aplicando transformada inversa:

\[
U=1-x^{-\alpha}
\]

\[
x^{-\alpha}=1-U
\]

\[
x=(1-U)^{-1/\alpha}
\]

Como \(1-U \sim U\):

\[
X=U^{-1/\alpha}
\]

```python
#pareto
alpha = 3

U = np.random.uniform(0,1,n)

X_pareto = U**(-1/alpha)

print("MEDIA:", np.mean(X_pareto))
print("VARIANZA:", np.var(X_pareto))
print("DESV. EST.:", np.std(X_pareto))

#hsitograma
x = np.linspace(1,6,500)

plt.figure(figsize=(8,5))
plt.hist(X_pareto, bins=60, density=True, alpha=0.6)

plt.plot(
    x,
    pareto.pdf(x, b=alpha),
    linewidth=2
)

plt.title("Histograma Pareto")
plt.xlabel("x")
plt.ylabel("densidad")
plt.grid()
plt.show()

#distribución acumulada
X_sorted = np.sort(X_pareto)
F_emp = np.arange(1,n+1)/n

plt.figure(figsize=(8,5))
plt.plot(X_sorted, F_emp, label="Empírica")

plt.plot(
    x,
    pareto.cdf(x, b=alpha),
    linewidth=2,
    label="Teórica"
)

plt.title("FDA Pareto")
plt.xlabel("x")
plt.ylabel("F(x)")
plt.legend()
plt.grid()
plt.show()

```
